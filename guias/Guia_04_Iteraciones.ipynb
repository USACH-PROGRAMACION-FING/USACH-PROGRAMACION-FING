> - No se usan `break` ni `continue`: una repetición siempre se corta **algorítmicamente**, por
>   la condición del `while`. Si el ciclo termina cuando encuentras algo, entonces «todavía no
>   lo encuentro» es parte del criterio de término y hay que escribirlo.
> - No se usa `while True`: la condición de un ciclo tiene que decir cuándo termina. Un ciclo
>   que no dice cuándo termina no se puede trazar.
> - El acumulador siempre parte en el **neutro de la operación**: 0 para sumar, 1 para
>   multiplicar.
