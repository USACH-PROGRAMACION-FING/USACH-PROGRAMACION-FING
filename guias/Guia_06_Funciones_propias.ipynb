>
> - Una función **devuelve** su resultado con `return`; no lo imprime. Quien la invoca decide

> - Una función no usa ni modifica variables definidas fuera de ella. Todo lo que necesita
>   llega por parámetros y todo lo que produce sale por `return`: así se entiende leyéndola
>   sola.
> - Toda función propia lleva un **docstring** entre triples comillas que diga qué hace, qué
>   recibe y qué entrega.
