# Strings

Los **`strings`** son un tipo de dato que permite almacenar textos. En particular, a diferencia de los tipos de datos numéricos, el string es un tipo de dato *compuesto*, es decir, se puede descomponer en partes. Sus partes fundamentales son los **caracteres**, que son una forma de representar grafemas (como letras, números y signos de puntuación). Por ejemplo, "a" es un grafema, o "!", o incluso el espacio entre caracteres, " ".

Para que el programa entienda algo como string, es necesario definirlos con comillas, ya sean 'simples' o "dobles", siempre siendo consistentes entre las comillas de apertura y las de cierre.

````python
# Ejemplo 1: Se definirá un string con comillas simples y luego con comillas dobles
print('hola mundo')
print("Hoy hay que estudiar")
````

Lo que se ilustró en el **Ejemplo 1** fue la impresión por pantalla de dos strings definidos, tanto con comillas simples, como con comillas dobles.

Si se quiere definir un **`string`** mezclando ambas comillas, arrojará un mensaje de error como lo muestran los **Ejemplos 2** y **3**:

````python
# Ejemplo 2: Se definirá un string que inicialmente comienza con comillas simples y
# termina con comillas dobles 
print('hola mundo")
````

````python
# Ejemplo 3: Se definirá un string que inicialmente comienza con comillas dobles y
# termina con comillas simples
print("hola mundo')
````

En ambos casos Python responde con un `SyntaxError`. El intérprete lee la comilla de apertura y busca una comilla **del mismo tipo** para cerrar; al no encontrarla, llega al final de la línea sin saber dónde termina el texto.

Este error aparece antes de que el programa se ejecute, porque Python no logra siquiera entender qué dice el código. Por eso, aunque la celda tuviera muchas líneas más, ninguna se ejecutaría.


Además de las comillas simples y dobles, Python ofrece otra forma de poder definir `string` de gran tamaño, llamados `docstring`. Ya se usaron en el apunte de funciones propias, para documentar lo que hace una función.

Este tipo de string se define con triple comillas.

El uso de un `docstring` se puede ver en el **Ejemplo 4**:

````python
# Ejemplo 4: Creación de un string multilínea o docstring
texto = """La idea de este tipo de strings es generar
documentación para que otros programadores puedan entender
lo que se está haciendo en este código"""

print(texto)
````

## Los strings son objetos

Los tipos de datos vistos hasta ahora se comportaban como valores sueltos: un `int` es un número y poco más se puede decir de él. Con los strings aparece algo distinto, y esa diferencia tiene nombre: en Python un string es un **objeto**.

Un objeto es un dato que viene acompañado de dos cosas:

*   **Atributos**: la información que el objeto guarda sobre sí mismo. En el caso de un string, sus caracteres y el orden en que están puestos.
*   **Métodos**: las operaciones que el objeto sabe hacer con su propio contenido. Un string sabe entregar su versión en mayúsculas, o responder si todos sus caracteres son dígitos. Un número no sabe hacer ninguna de esas dos cosas, porque sobre un número no tendrían sentido.

La diferencia entre una **función** y un **método** se nota al escribirlos. Una función recibe el dato entre paréntesis:

```python
len(palabra)
```

Un método, en cambio, se invoca *sobre* el dato, usando un punto:

```python
palabra.upper()
```

Ese punto se puede leer como "pídele a este objeto que…". Así, `palabra.upper()` se lee "pídele a `palabra` que entregue su versión en mayúsculas". Por eso `len()` funciona sobre muchos tipos de datos distintos, mientras que `.upper()` solo funciona sobre objetos que sepan hacer esa operación.

Los métodos propios de los strings se revisan más adelante, en la sección **Métodos**. Por ahora basta con quedarse con la idea: un string no es solo un texto guardado, es un texto que además sabe hacer cosas consigo mismo.


## Comillas dentro de un string

Las convenciones para nombrar variables ya se vieron en el apunte de introducción y aplican igual para los strings. Lo que sí es propio de este tipo de dato es qué ocurre cuando el texto que se quiere guardar contiene comillas.


Python permite hacer uso de comillas dentro de un `string`, si estas difieren con las utilizadas para definirlo. Lo anterior se refiere a que, si el string fue definido con comillas dobles, el texto que se quiere resaltar debe estar con comillas simples y viceversa.

````python
# Ejemplo 5: usar comillas de distinto tipo dentro del string
mensaje_1 = "Este es un cuadernillo de 'Google Colab'"
mensaje_2 = 'Se imprime un ejemplo para "comprender mejor"'

print(mensaje_1)
print(mensaje_2)
````

Del mismo modo, si quisiera usar las mismas comillas dentro del string y para definirlo, puedo hacerlo usando el **caracter de escape**, que corresponde a *backslash* (`\`), como se presenta en el Ejemplo 6.

````python
# Ejemplo 6: usar el carácter de escape
mensaje_1 = "Este es un cuadernillo de \"Google Colab\""
mensaje_2 = 'Se imprime un ejemplo para \'comprender mejor\''

print(mensaje_1)
print(mensaje_2)
````

## Operadores y Operaciones sobre Strings

Al igual que con los tipos de datos numéricos, en Python también existen **operadores** para trabajar con strings.

Se trata de los mismos símbolos **`+`** y **`*`** que sobre números representan la suma y la multiplicación. Sin embargo, cuando sus operandos son strings, pasan a significar otra cosa: **concatenación** ("pegar" dos strings) y **repetición** (crear varias copias del string, una detrás de otra).

### Concatenación

Esta operación consiste en **unir** dos o más cadenas de `strings` para formar una nueva cadena
```
string_1 + string_2 -> string_1string_2
```

El **Ejemplo 7** muestra lo indicado anteriormente:

````python
# Ejemplo 7: concatenación de dos strings

texto_1 = "hola"
texto_2 = "mundo"

nuevo_texto = texto_1 + texto_2
nuevo_texto_2 = texto_1 + " " + texto_2

print(nuevo_texto)
print(nuevo_texto_2)
````

Vale la pena destacar que la concatenación funciona únicamente si **ambos operandos** son del tipo string. Si se intenta unir un string con otro tipo de dato, como un `int` o un `float`, Python entrega un error.

Los Ejemplos 8 y 9 **generan un error de forma intencional**, y conviene leer con atención el mensaje de cada uno, porque no son iguales.





````python
# Ejemplo 8
'hola' + 3
````

````python
# Ejemplo 9
3 + 'hola'
````

Los dos ejemplos anteriores fallan, pero **con mensajes distintos**, y la diferencia depende de cuál es el primer operando:

*   `'hola' + 3` entrega `TypeError: can only concatenate str (not "int") to str`. Python leyó primero un string, entendió que la operación pedida era una concatenación, y se encontró con que lo que venía después no era texto.
*   `3 + 'hola'` entrega `TypeError: unsupported operand type(s) for +: 'int' and 'str'`. Aquí leyó primero un número, entendió que la operación era una suma, y no existe forma de sumar un número con un texto.

Es el mismo problema visto desde dos lados: el operador `+` necesita que ambos operandos sean del mismo tipo. Si se quiere unir un número a un texto, hay que convertirlo antes con `str()`.


### Repetición

Esta operación se realiza teniendo como primer operador un string y como segundo operador un número entero. Consiste en **repetir** un `string` tantas veces como indique el número por el cual esté siendo multiplicado:

```python
string_1 * 3 -> string_1string_1string_1
```

El **Ejemplo 10** muestra lo indicado anteriormente:

````python
# Ejemplo 10: repetición de un string
texto_1 = "hola"

nuevo_texto = texto_1 * 3
nuevo_texto_2 = 3 * texto_1

print(nuevo_texto)
print(nuevo_texto_2)
````

Adicionalmente, Python ofrece diversas funciones nativas que se pueden realizar con el tipo de dato `string`. Entre ellas vale destacar:
* **`len(<string>)`**: devuelve la cantidad de caracteres del string
* **`str(<expresion>)`**: devuelve la expresión ingresada en formato string

El **Ejemplo 11** ilustra lo mencionado anteriormente:

````python
# Ejemplo 11: uso de funciones de strings

oracion = "este es un ejemplo para calcular el largo"

largo = len(oracion)

numero = 66
texto_numero = str(numero)

numero_nuevo = numero*2
repeticion_numero = texto_numero*2

print(largo)
print(numero_nuevo)
print(repeticion_numero)
````

En el ejemplo anterior se calculó el largo o la cantidad de caracteres que componían el string **`oracion`**.

Posterior a ello, se transformó el número 66 a `string` y se realizó una comparación entre dos operaciones: multiplicación del número 66 por 2 y la multiplicación del número transformado a strings por 2 (repetición).

## Indexación

Del mismo modo que hay operadores para concatenar y repetir, existe una manera de acceder a los caracteres contenidos en el `string` de manera **individual**. 

Para lograr lo anterior, se tiene que llamar a la variable que contiene el `string` y acompañarlo de corchetes (`[ ]`) e indicar, dentro de estos, la posición numérica (contando de izquierda a derecha) de dicho caracter.

**Es importante tener en cuenta que el primer caracter siempre estará en la posición `0`, el segundo en la posición `1` y así sucesivamente.**

Tampoco olvidar que los espacios **también son considerados** al momento de recorrer un `string`, pues estos también son caracteres.

````python
# Ejemplo 12: Encontrar un caracter mediante su posición
palabra = "Hola Mundo"

# Se quiere mostrar por pantalla la letra "M", por lo que tendremos que indicar su posición
caracter = palabra[5]

print(caracter)
````

**IMPORTANTE:** Nunca se debe olvidar que el primer caracter se encuentra en la posición **0** y el último caracter se encuentra en la posición **largo - 1** o, en otras palabras, **`len(<string>) - 1`**.

Dicho esto, se puede acceder a cualquier posición existente dentro del 0 y el largo - 1. Si se accede a una posición fuera de esos rangos, Python entrega el error `IndexError: string index out of range`, el cual es uno de los más comunes que cometen los programadores novatos.

````python
# Ejemplo 13: encontrar un caracter mediante su posición
palabra = "Hola Mundo"

# Se quiere mostrar por pantalla la letra "M", por lo que tendremos que indicar su posición
caracter = palabra[25]

print(caracter)
````

## Inmutabilidad

La indexación permite **leer** cualquier caracter de un string. Lo que no permite es **cambiarlo**.

Los strings en Python son **inmutables**: una vez creados, su contenido no se puede modificar. Se pueden leer sus caracteres y se pueden construir strings nuevos a partir de ellos, pero el string original nunca cambia.

Si se intenta reemplazar un caracter usando su posición, Python entrega un error. La siguiente celda **genera un error de forma intencional**:


````python
# Ejemplo 14: un string no se puede modificar
palabra = "Hola Mundo"

palabra[0] = "B"
````

El mensaje `TypeError: 'str' object does not support item assignment` dice exactamente eso: un objeto `str` no admite que se le asigne un valor a una de sus posiciones.

Entonces, ¿cómo se cambia una letra? No se cambia: se **construye un string nuevo** a partir del original, usando las operaciones que ya se conocen.


````python
# Ejemplo 15: construir un string nuevo en lugar de modificar el original
palabra = "sol"

palabra_nueva = "S" + palabra[1] + palabra[2]

print(palabra)
print(palabra_nueva)
````

El string original sigue intacto: lo que se hizo fue armar otro a partir de sus caracteres. Más adelante, con *slicing*, se verá una forma mucho más corta de escribir esto mismo.

Hay un punto que suele confundir. Que los strings sean inmutables **no** significa que una variable no pueda cambiar de valor:


````python
# Ejemplo 16: la variable sí se puede reasignar
palabra = "Hola"
print(palabra)

palabra = "Chao"
print(palabra)
````

Aquí no se modificó ningún string. Lo que ocurrió es que la variable `palabra`, que antes apuntaba al string `"Hola"`, pasó a apuntar a otro string distinto, `"Chao"`. El string `"Hola"` nunca cambió: simplemente dejó de ser usado.

La distinción es importante y se va a repetir en el apunte de listas, donde ocurre justamente lo contrario: las listas **sí** se pueden modificar.


## Slicing

Además de acceder a una posición puntual, es posible acceder a un **tramo** del string. Para eso existe el operador *slice* (del inglés "rebanada"), que usa los mismos corchetes de la indexación, pero con hasta tres números separados por dos puntos:

```python
string[inicio:fin:paso]
```

*   `inicio`: primera posición que se incluye.
*   `fin`: posición en la que el corte se detiene, **sin incluirla**.
*   `paso`: de cuánto en cuánto se avanza. Es opcional y por defecto vale 1.

La regla del `fin` es la misma de `range()` en el apunte de iteración: el extremo derecho nunca se incluye. El resultado de un slice es siempre un **string nuevo**; el original, siendo inmutable, queda igual.


````python
# Ejemplo 17: slicing básico
palabra = "Hola Mundo"

# Posiciones 2 a 7 (la 8 no se incluye)
print(palabra[2:8])
````

Los tres números son opcionales. Omitir `inicio` significa "desde el principio", y omitir `fin` significa "hasta el final". Esto evita tener que calcular largos a mano:


````python
# Ejemplo 18: omitir los extremos
palabra = "Hola Mundo"

print(palabra[:4])    # desde el principio hasta la posición 4 (sin incluirla)
print(palabra[5:])    # desde la posición 5 hasta el final
print(palabra[:])     # el string completo
````

El tercer número, el `paso`, indica cada cuántas posiciones se toma un caracter:


````python
# Ejemplo 19: el paso
palabra = "Hola Mundo"

print(palabra[::2])     # de dos en dos, desde el principio hasta el final
print(palabra[0:5:2])   # de dos en dos, solo entre las posiciones 0 y 4
````

El paso también puede ser **negativo**, lo que hace que el recorrido vaya de derecha a izquierda. El caso más usado es `[::-1]`, que entrega el string invertido:


````python
# Ejemplo 20: paso negativo
palabra = "Hola Mundo"

print(palabra[::-1])
````

Conviene notar una diferencia práctica con la indexación: pedir una posición que no existe entrega un `IndexError`, pero pedir un **tramo** que se sale del string no falla, simplemente devuelve lo que alcance a haber. `"Hola"[2:100]` entrega `"la"` sin quejarse.


## Métodos

Como se vio al comienzo del apunte, un string es un **objeto**, y por lo tanto trae consigo un conjunto de **métodos**: operaciones que sabe realizar sobre su propio contenido. Esta sección revisa los más usados.

Todos se invocan con la notación de punto, sobre el string mismo o sobre la variable que lo contiene.


Algunos métodos interesantes de los string que podemos usar son (en esta notación, cuando decimos `<string>`, queremos decir que desde `<` hasta `>` se debe reemplazar por el dato a utilizar):

*   `<string>.lower()`: retorna una copia del string donde todas sus letras son minúsculas.
*   `<string>.upper()`: retorna una copia del string donde todas sus letras son mayúsculas.
*   `<string>.title()`: retorna una copia del string donde la primera letra de cada palabra de un string está en mayúscula.
*   `<string>.swapcase()`: retorna una copia del string donde cada minúscula es convertida a mayúscula y viceversa.
*   `<string>.islower()`: retorna `True` si los caracteres del strings son solo letras minúsculas.
*   `<string>.isupper()`: retorna `True` si los caracteres del strings son solo letras mayúsculas.
*   `<string>.isdigit()`: retorna `True` si los caracteres del strings son solo numéricos.
*   `<string>.isalpha()`: retorna `True` si los caracteres del strings son solo letras.

**MANOS A LA OBRA:** 
* Revisa qué ocurre con `.isdigit()` si el string representa un número negativo, como por ejemplo `-32`, o un número flotante como `3.1456`.
* Revisa qué ocurre con `.isalpha()` si el string tiene tildes.

````python
# Ejemplo 21: método .lower() y .upper()
palabra = "Hola Mundo"

palabra_minuscula = palabra.lower()
palabra_mayuscula = palabra.upper()

print(palabra_minuscula)
print(palabra_mayuscula)
print(palabra)
````

Vale la pena detenerse en algo: los métodos de los strings **no modifican el string original**, sino que entregan una copia con la transformación aplicada. Esto no es una decisión caprichosa, es una consecuencia directa de la **inmutabilidad** vista más arriba: como el string original no se puede cambiar, lo único que un método puede hacer es construir uno nuevo.

En el Ejemplo 21 se ve que quedan tres variables con tres strings distintos, y que el original sigue intacto. El Ejemplo 22 muestra cómo aprovechar que las variables sí se pueden reasignar, para aplicar transformaciones sucesivas.


````python
# Ejemplo 22: método .title() y .swapcase()
palabra = "hola mundo"

palabra_titulo = palabra.title()
palabra_cambiada = palabra_titulo.swapcase()

print(palabra_titulo)
print(palabra_cambiada)
````

````python
# Ejemplo 23: métodos .islower() e .isupper()
palabra = "Hola Mundo"
palabra_2 = "HOLA MUNDO"
palabra_3 = "hola mundo"

palabra_minuscula = palabra.islower()
palabra_minuscula_2 = palabra_3.islower()

palabra_mayuscula = palabra.isupper()
palabra_mayuscula_2 = palabra_2.isupper()

print(palabra_minuscula)
print(palabra_minuscula_2)

print(palabra_mayuscula)
print(palabra_mayuscula_2)
````

````python
# Ejemplo 24: métodos .isalpha() y .isdigit()
palabra = "HolaMundo"
palabra_2 = "espero que pronto sea 31 de diciembre"
palabra_3 = "15842"

palabra_letra = palabra.isalpha()
palabra_letra_2 = palabra_2.isalpha()

palabra_numero = palabra_2.isdigit()
palabra_numero_2 = palabra_3.isdigit()

print(palabra_letra)
print(palabra_letra_2)

print(palabra_numero)
print(palabra_numero_2)
````

## Recorrido de strings

Un string está compuesto por caracteres, y muchas veces el problema a resolver exige pasar por **todos** ellos: contar cuántas vocales tiene una palabra, revisar si un RUT viene con puntos y guion, transformar letra por letra. A eso se le llama **recorrer** el string.

Existen tres formas de hacerlo, y las tres usan herramientas del apunte de iteración.

### Forma 1: `while` con un contador

La variable `i` lleva la cuenta de la posición actual y el ciclo avanza mientras esa posición siga existiendo, es decir, mientras sea menor que el largo del string.


````python
# Ejemplo 25: recorrido con while
texto = "Ingeniería"

i = 0
while i < len(texto):
  print(i, texto[i])
  i = i + 1
````

La condición es `i < len(texto)` y no `i <= len(texto)`, por la misma razón vista en indexación: la última posición válida es `len(texto) - 1`. Usar `<=` provocaría un `IndexError` en la última vuelta.

### Forma 2: `for-in` directo

Si no interesa la posición sino solo el caracter, el `for-in` recorre el string sin necesidad de contador:


````python
# Ejemplo 26: recorrido con for-in
texto = "Ingeniería"

for caracter in texto:
  print(caracter)
````

### Forma 3: `for-in` sobre `range(len(texto))`

Cuando se necesita el caracter **y además** su posición, se recorre un `range()` construido con el largo del string:


````python
# Ejemplo 27: recorrido con for-in y range(len())
texto = "Ingeniería"

for i in range(len(texto)):
  print(i, texto[i])
````

### ¿Cuál usar?

Las tres recorren el mismo string, pero no sirven para lo mismo:

*   **`for-in` directo**: cuando solo importa el caracter. Es la forma más corta y en la que menos se puede fallar, porque no hay contador que actualizar ni condición que ajustar.
*   **`for-in` con `range(len())`**: cuando además de cada caracter se necesita saber en qué posición está.
*   **`while`**: cuando el recorrido no va de principio a fin de uno en uno. Por ejemplo, si hay que detenerse apenas se encuentra algo, saltarse caracteres según una condición, o avanzar de a dos en ciertos casos y de a uno en otros.

La recomendación del curso es partir siempre por el `for-in` directo y recurrir a las otras dos formas solo cuando el problema lo exija.

### Sobre el nombre del identificador

Al recorrer con `for-in`, la unidad del recorrido la decide **el tipo de dato**, no el nombre que se le ponga al identificador. El siguiente ejemplo lo deja en evidencia:


````python
# Ejemplo 28: el nombre del identificador no cambia la unidad del recorrido
texto = "Esto es Python"

for palabra in texto:
  print(palabra)
````

El nombre `palabra` hace pensar que el ciclo entregará `Esto`, `es` y `Python`. Pero al recorrer un `string`, Python siempre entrega **un caracter a la vez**, se llame como se llame el identificador. Recorrer por palabras requiere una herramienta distinta, que se verá en el apunte de listas.

Para evitar esa confusión, conviene que el identificador nombre la unidad real del recorrido:

*   Objeto iterable **string** → identificador `caracter`
*   Objeto iterable **lista** → identificador `elemento`
*   Objeto iterable **archivo** → identificador `linea`

### Un ejemplo completo

El siguiente programa junta casi todo lo visto en el apunte: pide una palabra, la pasa a minúsculas con un método, la recorre con `for-in` y usa el operador de pertenencia `in` del apunte de decisiones para contar sus vocales.


````python
# Ejemplo 29: contar las vocales de una palabra
# CONSTANTES
VOCALES = "aeiouáéíóú"

# ENTRADA
palabra = input("Ingresa una palabra: ")

# PROCESAMIENTO
contador = 0
for caracter in palabra.lower():
  if caracter in VOCALES:
    contador = contador + 1

# SALIDA
print("La palabra", palabra, "tiene", contador, "vocales")
````
