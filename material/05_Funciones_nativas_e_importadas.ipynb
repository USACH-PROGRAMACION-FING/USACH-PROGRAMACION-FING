# Funciones nativas e importadas

Las funciones, en programación, corresponden a porciones de código aislado que realizan una determinada tarea. 

Esto implica que una función es una especie de *"mini-programa"*, el cual a partir de alguna(s) entradas, entrega una salida.

En Python existen una gran cantidad de funciones predefinidas y listas para usarse. Estas se llaman **funciones nativas**, dado que son propias del lenguaje de programación. En otras palabras, Python automáticamente entiende a qué nos referimos cuando usamos una de ellas, como por ejemplo, `print()`, `len()` o `input()`.

En general, estas funciones resuelven problemas específicos y comunes a la hora de programar, como por ejemplo, redondear un número o mostrar un mensaje en la pantalla del usuario. Sin embargo, las funciones nativas no bastan por sí solas para resolver problemas específicos. En estos casos, tenemos dos opciones: 
* Importar funciones que alguien más haya creado, o bien,
* Definir nuestras propias funciones.

Una de las grandes ventajas de Python es que, dada su popularidad, distintos usuarios han desarrollado múltiples funciones para resolver problemas en particular. Normalmente cuando se crea un conjunto de funciones para resolver algún problema en particular, como por ejemplo, hacer un programa con conexión a internet, estas se empaquetan en una *biblioteca* o **módulo**, y luego se dejan disponibles para que la comunidad pueda usarlas.

Existen algunos módulos que fueron creados por los mismos desarrolladores de Python, los cuales podemos **importar** directamente en nuestros programas. Pero también existen módulos desarrollados por otros usuarios y organizaciones (terceros, o *third-parties*) los que deben instalarse para poder usarse. Sin embargo, todas estas funciones se conocen como **funciones importadas**, pues debo "importarlas" o *"traerlas"* a mi código para poder usarlas.

Tanto las funciones nativas como importadas trabajan bajo el concepto de **caja negra**, es decir, podemos usarlas solo conociendo:
* El **nombre** que las identifica.
* Los parámetros de **entrada** que necesita la función para poder operar (hay funciones que necesitan uno, otras ninguno y otras tienen múltiples parámetros).
* La **salida** o el tipo de salida que generan.

Conociendo esto, puedo **invocar** a la función, la cual ejecutará un **proceso** encapsulado, es decir, ejecutan una serie de instrucciones hasta obtener una **salida**.

Dado que la función siempre ejecutará el proceso, no es necesariamente relevante poder saber **cómo** ejecuta el proceso, sino que solo basta saber **qué** es lo que hace. 

Por ejemplo: ¿vale la pena saber qué es lo que hace internamente un computador cuando uso un `input()`?

* En general no, pues nos basta con saber *qué hace* la función y que la podemos usar. De hecho, la hemos usado todo el curso, sin saber cómo funciona internamente.

## Invocación de funciones

Antes de revisar qué funciones ofrece Python, conviene detenerse en **cómo** se usan. Llamar a una función se dice **invocarla**, y su sintaxis es siempre la misma:

```python
nombre_funcion(argumento_1, argumento_2, ...)
```

*   El **nombre** identifica qué función se quiere usar.
*   Los **paréntesis** son obligatorios, incluso cuando la función no recibe nada, como en `random()`. Sin ellos, Python entiende que te refieres a la función, no que quieres ejecutarla.
*   Dentro de los paréntesis van los **argumentos**, separados por comas: los datos con los que la función va a trabajar. Cuántos son y de qué tipo depende de cada función.

Ya has invocado varias durante el curso: `print("hola")`, `input("Ingresa un número: ")`, `int(texto)`. En todas se repite el mismo patrón.

Hay una diferencia entre ellas que conviene tener clara. La mayoría de las funciones **devuelven** un valor, y ese valor hay que guardarlo en una variable o usarlo de inmediato; si no, se calcula y se pierde:


````python
# Ejemplo 1: el valor que devuelve una función hay que guardarlo o usarlo
numero = -7

abs(numero)                   # el resultado se calcula y se pierde
valor_absoluto = abs(numero)  # el resultado queda guardado en una variable

print(valor_absoluto)
````

Un caso distinto es `print()`, que no devuelve un valor útil: su efecto es mostrar algo por pantalla. Por eso `print()` se escribe como una línea completa y no tiene sentido guardarlo en una variable.


## Funciones Nativas

Las **funciones nativas** (en inglés, *built-in functions*) corresponden a las funciones que están **por defecto en el lenguaje Python.**

Estas funciones pueden ser invocadas de manera directa y usarse en el programa sin darles una definición previa.

Durante el transcurso del curso, se han visto diversas funciones que ayudan al proceso de buscar la solución a un determinado problema. Existen funciones que sirven para temas generales y otras para operaciones más específicas.

Un ejemplo de estas funciones es `len()`, cuyo nombre proviene del inglés "*length*". Esta es, precisamente, una **función nativa** que permite contar la cantidad de caracteres de un string, es decir, su largo ("*its length*"). Más adelante se verá que también funciona sobre otros tipos de datos compuestos. Para usarla no hay que importar ni definir nada, basta con invocarla: `len(palabra)`.

A continuación se presentan algunos ejemplos de funciones nativas:


|Nombre        |Entrada                 | Salida     | Proceso              |
|:------------:|:-----------------------|------------|----------------------| 
|`int(x)`      |Un número o string  numérico|Un número|Devuelve un número entero| 
|`float(x)`    |Un número o string  numérico|Un número|Devuelve un número flotante|
|`bool(x)`     |Un dato|Un booleano|Devuelve `False` si el dato es vacío, de lo contrario entrega `True`|
|`str(x)`      |Un dato|Un string|Convierte el dato en un string|
|`abs(x)`      |Un número|Un número|Devuelve el valor absoluto de un número|
|`pow(x, y)`   |Dos números|Un número|Equivale a realizar x**y|
|`len(x)`      |Un string|Un entero|Devuelve la cantidad de caracteres que contiene|
|`max(x, y, ...)`|Dos o más números|Un número|Devuelve el mayor de los valores recibidos|
|`min(x, y, ...)`|Dos o más números|Un número|Devuelve el menor de los valores recibidos|
|`round(x, y)` |Un flotante (x) y un entero (y)|Un flotante|Devuelve el número x redondeado a la cantidad de decimales indicada (y)

````python
# Ejemplo 2: conversiones de tipo
# Entrada
numero_string = "45"

# Procesamiento
numero_entero = int(numero_string)
numero_flotante = float(numero_string)
booleano = bool(numero_string)

# Salida
print("número entero", numero_entero)
print("número flotante", numero_flotante)
print("booleano", booleano)
````

````python
# Ejemplo 3: convertir a texto para poder medir su largo
# Entrada
numero = 123456789

# Procesamiento
# Un número no tiene "largo", pero su representación como texto sí
numero_texto = str(numero)
cantidad_digitos = len(numero_texto)

# Salida
print("El número", numero, "tiene", cantidad_digitos, "dígitos")
````

````python
# Ejemplo 4: valor absoluto y potencia
# Entrada
numero = -5

# Procesamiento
numero_absoluto = abs(numero)
exponente = pow(numero_absoluto, 3)

# Salida
print("valor absoluto", numero_absoluto)
print("potencia", exponente)
````

````python
# Ejemplo 5: el mayor y el menor de varios valores
# Entrada
nota_1 = 5.4
nota_2 = 3.8
nota_3 = 6.7

# Procesamiento
nota_mas_baja = min(nota_1, nota_2, nota_3)
nota_mas_alta = max(nota_1, nota_2, nota_3)

# Salida
print("nota más baja", nota_mas_baja)
print("nota más alta", nota_mas_alta)
````

````python
# Ejemplo 6: redondeo
# Entrada
numero = 25.4627

# Procesamiento
redondear = round(numero, 3)
# Si no se da el segundo parámetro, redondea a entero
numero_entero = round(numero)

# Salida
print("número original", numero)
print("número redondeado", redondear)
print("número redondeado a entero", numero_entero)
````

El cuadro anterior describe solo algunas de las funciones nativas que Python posee. Para poder conocer una mayor cantidad de este tipo de funciones, se recomienda visitar el siguiente enlace: [Funciones Nativas](https://docs.python.org/es/3.10/library/functions.html)

Para ilustrar la ventaja de usar algunas funciones nativas de Python, se presenta el siguiente ejercicio, el cual se resuelve de dos maneras: en primer lugar, usando iteración y luego aprovechando algunas funciones nativas.

### Ejercicio

Se solicita calcular la suma de los primeros $n$ números naturales: el mismo problema que se resolvió con un ciclo en el apunte de iteración.

````python
# Ejemplo 7: solución utilizando un ciclo
# Entrada
numero_final = 9

# Procesamiento
i = 1
suma = 0
while i <= numero_final:
  suma = suma + i
  i = i + 1

# Salida
print("La suma de los primeros", numero_final, "números naturales es", suma)
````

````python
# Ejemplo 8: la misma solución, usando una función nativa
# Entrada
numero_final = 9

# Procesamiento
suma = sum(range(1, numero_final + 1))

# Salida
print("La suma de los primeros", numero_final, "números naturales es", suma)
````

Los dos programas entregan el mismo resultado. El primero construye la suma paso a paso; el segundo le entrega a la función nativa `sum()` el mismo `range()` que el ciclo habría recorrido, y deja que ella haga el recorrido.

Esa es la ventaja de conocer las funciones nativas: no evitan pensar el problema —hay que darse cuenta de que los números a sumar son exactamente `range(1, n + 1)`—, pero una vez planteado, ahorran escribir el ciclo y, con él, la posibilidad de equivocarse en la condición o en la actualización del contador.


## Funciones Importadas

Dado que las funciones nativas resuelven problemas más o menos generales, para problemas más específicos, es necesario saber cómo trabajar con **módulos**.

Las funciones de estos módulos, si bien vienen instaladas junto a Python, no son reconocidas por este automáticamente para poder usarse, por lo que debemos acceder **explícitamente al módulo**  para poder usarlas (esto es así, pues Python es un lenguaje de programación general, bien podríamos hacer un programa para calcular impuestos, como un videojuego: ¿qué sentido tiene tener cargados en la memoria elementos para interfaces gráficas y manejo de elementos 3D, si solo necesitamos operaciones matemáticas sencillas para el cálculo de los impuestos? Sería un desperdicio de recursos).

Un módulo indispensable para un estudiante de ingeniería es `math`, el cual contiene muchas de las funciones matemáticas y trigonométricas que se requieren usualmente.

Algunas de las funciones que se pueden encontrar en el módulo `math` son la siguientes:

|Nombre        |Descripción |
|:------------:|:-----------| 
|`sin(x)`|Obtiene el seno de un ángulo x expresado en radianes|
|`cos(x)`|Obtiene el coseno de un ángulo x expresado en radianes|
|`tan(x)`|Obtiene la tangente de un ángulo x expresado en radianes|
|`exp(x)`|Obtiene el número e elevado a x|
|`log(x)`|Obtiene el logaritmo natural (base $e$) de x|
|`log10(x)`|Obtiene el logaritmo en base 10 de x|
|`sqrt(x)`|Obtiene la raíz cuadrada de x|
|`degrees(x)`|Convierte a grados un ángulo x expresado en radianes|
|`radians(x)`|Convierte a radianes un ángulo x expresado en grados|

A continuación, se presenta un ejemplo donde se calculará el coseno de un ángulo dado por el usuario. El ejemplo **genera un error de forma intencional**: el código está bien escrito, pero le falta algo.

````python
# Ejemplo 9: esta celda genera un error de forma intencional
# Entrada
angulo = int(input("Favor ingresar ángulo en grados: "))

# Procesamiento
# Se trasforman los grados a radianes
radianes = radians(angulo)

# calcular coseno del ángulo
coseno = cos(radianes)

# Salida
print("El coseno del ángulo", angulo, "es de", coseno)
````

El error del **Ejemplo 9** es `NameError: name 'radians' is not defined`, y se produce porque no se le ha indicado a Python que tanto `radians()` como `cos()` pertenecen al módulo `math`, que debe importarse. Para Python, `radians` es sencillamente un nombre que nadie definió.

### Importar Funciones

Para realizar la importación de funciones se pueden utilizar 4 alternativas:

1.   Importar solo el módulo ✔️
2.   Importar el módulo con un alias ✔️
3.   Importar solo las funciones requeridas del módulo ✔️
4.   Importar todas las funciones del módulo ❌

#### 1. Importar solo el módulo

Al realizar la importación de esta manera, se pueden utilizar todas las funciones del módulo importado siempre que se haga referencia al módulo de origen al momento de utilizar la función. Esto se consigue mediante la utilización de la palabra reservada **`import`**.

```python
import <modulo>

<modulo>.funcion(<parametros>)
```

Para ilustrar este uso, se corrige el programa del ejemplo anterior.

````python
# Ejemplo 10: importar el módulo completo
# Importación de Funciones
import math

# Entrada
angulo = int(input("Favor ingresar ángulo en grados: "))

# Procesamiento
# Se trasforman los grados a radianes
radianes = math.radians(angulo)

# Calcular coseno del ángulo
coseno = math.cos(radianes)

# Salida
print("El coseno del ángulo", angulo, "es de", coseno)
````

#### 2. Importar el módulo con un alias

En este tipo de importación, se utiliza un alias ("sobrenombre") para llamar al módulo importado. De esta manera, se pueden utilizar todas las funciones siempre y cuando, al momento de llamarlas, se anteponga el alias a cada función. Esto se consigue mediante la utilización de la palabra reservada **`import`** en combinación con **`as`**.

```
import <modulo> as <alias>

<alias>.funcion(<parametro>)
```

Usando el mismo ejemplo:

````python
# Ejemplo 11: importar el módulo con un alias
# Importación de Funciones
import math as m

# Entrada
angulo = int(input("Favor ingresar ángulo en grados: "))

# Procesamiento
# Se trasforman los grados a radianes
radianes = m.radians(angulo)

# Calcular coseno del ángulo
coseno = m.cos(radianes)

# Salida
print("El coseno del ángulo", angulo, "es de", coseno)
````

Los alias sirven especialmente en ocasiones en las que el nombre de un módulo es muy grande, estamos importando un submódulo o hacemos referencia a algo de manera muy seguida. Existen alias que la comunidad usa por convención y que conviene reconocer. El del ejemplo anterior, `math as m`, **no** es uno de ellos: `math` ya es un nombre corto y casi nunca se abrevia. Los alias realmente establecidos son otros, como `numpy as np` o `pandas as pd`. Otro caso clásico es el módulo para realizar gráficos, `matplotlib` (que no viene con Python y debe instalarse aparte), que tiene un submódulo llamado `pyplot` que define las principales interfaces y funciones para graficar, se suele abreviar con la línea:
```python
import matplotlib.pyplot as plt
```

#### 3. Importar solo las funciones requeridas del módulo

Al realizar la importación de esta manera, no se requiere hacer mención al módulo de origen. No obstante, solo se pueden utilizar directamente las funciones indicadas al momento de realizar la importación. Esto se consigue mediante la utilización de las palabras reservadas **`from`** e **`import`**.

```
from <modulo> import <función1>, <función2>, <función3,...>
```

En este caso, el ejemplo original quedaría como:

````python
# Ejemplo 12: importar solo las funciones requeridas
# Importación de Funciones
from math import radians, cos

# Entrada
angulo = int(input("Favor ingresar ángulo en grados: "))

# Procesamiento
# Se trasforman los grados a radianes
radianes = radians(angulo)

# Calcular coseno del ángulo
coseno = cos(radianes)

# Salida
print("El coseno del ángulo", angulo, "es de", coseno)
````

Esta forma de importar debe ser utilizada con cuidado, pues, como se ve, las funciones son llamadas como si fuesen funciones nativas y si no tenemos cuidado, podemos reemplazar variables locales u otras importaciones.

> ### ⚠️ Mala práctica en este curso: `from <modulo> import *`
>
> La cuarta forma de importar trae **todas** las funciones de un módulo de una vez, y permite usarlas directamente, sin anteponer nada:
>
> ```python
> from math import *
>
> radianes = radians(angulo)
> coseno = cos(radianes)
> ```
>
> Funciona, y a primera vista parece la más cómoda de las cuatro. **En este curso no se va a utilizar**, y fuera del curso también se considera una mala práctica.
>
> El problema es que no queda registro de qué nombres se trajeron. El módulo `math` define varias decenas de funciones y constantes, y con `from math import *` todas quedan sueltas dentro del programa. Si más adelante defines una variable llamada `pi`, `e` o `log`, la estarás pisando sin darte cuenta. Y si importas dos módulos con `*`, el segundo sobrescribe en silencio cualquier nombre que compartiera con el primero.
>
> Además, quien lea el programa no tiene cómo saber de dónde viene `cos()`: tendría que conocerse de memoria el contenido de cada módulo importado.
>
> Las tres formas anteriores no tienen ese problema, porque en todas queda escrito qué se está trayendo: `import math` mantiene el nombre del módulo a la vista en cada uso, y `from math import radians, cos` nombra exactamente las dos funciones que se van a usar.


### Módulo `random`

Otro módulo útil que Python trae por defecto es **`random`**, el cual permite la generación de números pseudoaleatorios (es decir, que se genera con un método que *parece* aleatorio).

Alguna de las funciones que podemos usar se presentan a continuación:

|Nombre        |Descripción |
|:------------:|:-----------| 
|`randint(x, y)`|Devuelve un numero entero pseudoaleatorio entre los valores indicados (ambos valores inclusive)|
|`random()`|Devuelve un número flotante pseudoaleatorio entre 0 y 1, sin incluir el 1|

````python
# Ejemplo 13: números pseudoaleatorios
# Importación de Funciones
from random import randint, random

# Constantes
VALOR_MINIMO = 3
VALOR_MAXIMO = 10

# Procesamiento
numero_entero = randint(VALOR_MINIMO, VALOR_MAXIMO)
numero_flotante = random()

# Salida
print("entero pseudoaleatorio entre", VALOR_MINIMO, "y", VALOR_MAXIMO, ":", numero_entero)
print("flotante pseudoaleatorio entre 0 y 1:", numero_flotante)
````

Cada vez que se ejecuta la celda anterior, los valores cambian. Combinando `randint()` con un ciclo del apunte de iteración se puede simular un experimento repetido, como lanzar un dado:


````python
# Ejemplo 14: simular el lanzamiento de un dado
# Importación de Funciones
from random import randint

# Constantes
CARA_MINIMA = 1
CARA_MAXIMA = 6
LANZAMIENTOS = 10

# Procesamiento y salida
i = 1
while i <= LANZAMIENTOS:
  print("Lanzamiento", i, ":", randint(CARA_MINIMA, CARA_MAXIMA))
  i = i + 1
````

El módulo `random` incluye además `choice()`, que escoge un elemento al azar, y `shuffle()`, que desordena los elementos. Ambas trabajan sobre datos **compuestos**, así que se podrán usar a partir de los apuntes de strings y, sobre todo, de listas.


Usamos el término *pseudoaleatorio* y no aleatorio, debido a que, cuando un computador ejecuta un proceso (aún el de escoger un número *al azar*), este comportamiento no puede ser aleatorio. Dado que los computadores son dispositivos determinísticos, es decir, su comportamiento es **completamente predecible**, en la práctica resulta extremadamente difícil que hagan una elección al azar.

Por ello, los procesos que generan números pseudoaleatorios, en la práctica son algoritmos matemáticos cuyas salidas son *suficientemente difíciles* de predecir, pero que, si tenemos suficientes datos, podemos predecir con absoluta certeza. Como la aleatoriedad es muy importante, por ejemplo, en criptografía, siempre se están usando técnicas para aumentar la aleatoriedad de los valores, como por ejemplo, el mecanismo que utiliza [Cloudflare](https://www.cloudflare.com/es-es/learning/ssl/lava-lamp-encryption/).

```python
import matplotlib.pyplot as plt
```

Dos cosas que conviene saber:
